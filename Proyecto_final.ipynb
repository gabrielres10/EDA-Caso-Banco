Una entidad financiera requiere la elaboración de un modelo de clasificación que permita identificar el nivel de riesgo de lavado de activos y financiación del terrorismo (LAFT), que poseen sus clientes, puesto que de acuerdo con la Circular Básica Jurídica (C.E 029/14) en su Parte 1 – Título IV – Capitulo IV: Instrucciones relativas a la administración del riesgo de lavado de activos y de la financiación del terrorismo – SARLAFT, se establece qué:

_“[…] en aquellos eventos en que hayan identificado que el perfil de riesgo de un potencial cliente esté calificado como de alto riesgo, las entidades vigiladas deben aplicar medidas intensificadas en los procedimientos de conocimiento del cliente. […]”_

Por lo anterior, es necesario que las entidades financieras conozcan el nivel de riesgo de LAFT que representan sus clientes.

Para esto, dicha entidad ha realizado la entrega de dos archivos, los cuales contienen, la información básica de vinculación de los clientes y su información transaccional con fecha de corte al 21 de febrero de 2025, dentro de esta información se evidencia una columna en la cual se esboza el nivel de riesgo LAFT de los clientes.

```python
risk_form = data['ACRM - NIVEL DE RIESGO'].value_counts(normalize=True).rename('Proportion').reset_index()
risk_form.rename(columns={risk_form.columns[0]: 'Nivel de Riesgo', 'Proportion': 'Porcentaje'}, inplace=True)
risk_form['Porcentaje'] = np.round(risk_form['Porcentaje'] * 100, decimals=2)

if risk_form['Nivel de Riesgo'].dtype in ['int64', 'float64']:
    risk_form['Nivel de Riesgo'] = risk_form['Nivel de Riesgo'].astype(int)

palette = {'Medio': 'orange', 'Alto': 'red', 'Bajo': 'green'}
bar = sns.barplot(data=risk_form, x='Nivel de Riesgo', y='Porcentaje', hue='Nivel de Riesgo', palette=palette, legend=False)  # Cambios aquí

for i in range(risk_form.shape[0]):
    bar.text(
        i, risk_form.loc[i, 'Porcentaje'] / 3,
        f"{risk_form.loc[i, 'Porcentaje']}%\n{data['ACRM - NIVEL DE RIESGO'].value_counts()[risk_form.loc[i, 'Nivel de Riesgo']]}",
        fontdict=dict(color='white', fontsize=12, weight='bold'),
        horizontalalignment='center'
    )
```

# Depuración de variables redundantes y de valores duplicados

Ahora que ya tenemos todas las columnas numéricas, podemos hacer algunos análisis que antes no se podían realizar por la naturaleza de los datos. Primero, veamos su distribución

```python
numeric_data.describe().T
```

Vamos a analizar la correlación que tienen las variables para escoger las mejores dentro del conjunto de datos. Esto nos permitirá evitar repetir información, ya que muchas veces algunas columnas son tan dependientes entre sí que prácticamente proporciona la misma información

```python
plt.figure(figsize=(25, 20))
sns.heatmap(numeric_data.corr(), annot=False, cmap="coolwarm", linewidths=0.5)
plt.show()
```

Antes del análisis de correlación, se ven varios puntos rojos en el gráfico, que representan variables bastante correlacionadas que proporcionan información muy parecida. Para evitar tener datos redundantes, vamos a emplear un método para eliminar las variables que tengan más de 40% de correlación entre sí, con el fin de eliminar redundancia.

```python
non_redundant_df = numeric_data.copy()

# Definir un umbral de correlación (ejemplo: 0.4)
correlation_threshold = 0.4

# Calculamos la matriz de correlación
corr_matrix = non_redundant_df.corr().abs()

# Seleccionamos las variables con alta correlación
upper_triangle = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))

# Encontramos las columnas a eliminar (si tienen una correlación mayor al umbral)
to_drop = [column for column in upper_triangle.columns if any(upper_triangle[column] > correlation_threshold)]

# Eliminamos las columnas seleccionadas
non_redundant_df = non_redundant_df.drop(columns=to_drop)

non_redundant_df['NIVEL DE RIESGO'] = numeric_data['ACRM - NIVEL DE RIESGO']
non_redundant_df = non_redundant_df.drop(columns=['ACRM - SCORE PRODUCTO'])

# Mostramos las variables eliminadas
print("Variables eliminadas por alta correlación:", to_drop)
```

Parece que se han eliminado múltiples columnas debido a su alta correlación. Esto debería verse reflejado en el mapa de calor. A continuación, veamos el resultado:

```python
plt.figure(figsize=(25, 20))
sns.heatmap(non_redundant_df.corr(), annot=False, cmap="coolwarm", linewidths=0.5)
plt.show()
```

Con este gráfico, podemos observar todas las columnas que quedaron después de eliminar las redundancias. Se puede apreciar que ya no hay una cantidad tan grande de variables dependientes entre sí.

No obstante, ambos gráficos poseen unos espacios vacíos en su interior, lo que sugiere que algunas de las variables tienen un solo valor en todos sus registros. Esto es un problema porque realmente no aportan nada de información que explique la variable objetivo y solo aumenta el tamaño de los datos. Por este motivo, el siguiente método busca todas las columnas que tengan solo un valor en todos sus registros y las elimina.

```python
# Encontrar columnas con un solo valor único
columns_one_value = [col for col in non_redundant_df.columns if non_redundant_df[col].nunique() == 1]

# Mostrar las columnas detectadas
print("Columnas con un solo valor:", columns_one_value)
print(len(columns_one_value))

non_redundant_df = non_redundant_df.drop(columns=columns_one_value)
```

Ahora podemos verificar cómo quedaría el mapa de calor sin las columnas con un solo valor en todos sus registros

```python
plt.figure(figsize=(25, 20))
sns.heatmap(non_redundant_df.corr(), annot=False, cmap="coolwarm", linewidths=0.5)
plt.show()
```

```python
# Tabla de contingencia entre las variables de interés
tabla_contingencia = pd.crosstab(data['ACRM - NIVEL DE RIESGO'], data['NIVEL RIESGO ACT ECONOMICA'])

# Mostrar la tabla
print(tabla_contingencia)
```

```python
# Frecuencias relativas (proporción sobre el total)
tabla_contingencia_prop = pd.crosstab(data['ACRM - NIVEL DE RIESGO'], data['NIVEL RIESGO ACT ECONOMICA'], normalize=True)
print(tabla_contingencia_prop)
```

**Análisis de la Relación Entre Variables basado en la tabla de contingencia**  
  
A partir de las tablas de contingencia entre las frecuencias y las proporciones de estas dos columnas evaluadas, podemos concluir lo siguiente:

- La mayoría de las personas con nivel de riesgo BAJO en actividades económicas también tienen un nivel de riesgo bajo de actividades ilícitas.  
- Los casos de alto riesgo de activiades ilícitas están más distribuidos entre las categorías de actividad económica, pero siguen siendo pocos.  


Con esta nueva información, parece haber una correlación **moderada** entre ambas variables.


Vamos a analizar otro de los indicadores estadísticos. El índice **V de Cramér** se usa para medir la **fuerza de asociación** entre dos variables categóricas (0 = sin relación, 1 = relación fuerte).  

```python
import scipy.stats as stats
import numpy as np

def cramers_v(tabla):
    chi2 = stats.chi2_contingency(tabla)[0]
    n = tabla.sum().sum()  # Total de observaciones
    k = min(tabla.shape) - 1  # Mínimo entre filas y columnas - 1
    return np.sqrt(chi2 / (n * k))

# Calcular V de Cramér
v_cramer = cramers_v(tabla_contingencia)
print(f'V de Cramér: {v_cramer:.4f}')
```

**Interpretación:**  
- Contrario a lo que habíamos planteado al inicio, podemos decir que un valor de **0.2577** indica una **asociación moderada** entre ambas variables, por lo que la relación entre estas variables parece no ser tan alta como pensabamos.  
- **No es una relación muy fuerte, pero sí hay cierta dependencia** entre el riesgo de la actividad económica y el nivel de riesgo en ACRM.  


Realizaremos un análisis utilizando otras medidas de asociación que nos darán una pista más certera sobre la existencia de esta relación


```python
from scipy.stats import chi2_contingency

# Crear una tabla de contingencia
tabla_contingencia_prop = pd.crosstab(data['ACRM - NIVEL DE RIESGO'], data['NIVEL RIESGO ACT ECONOMICA'])

# Realizar la prueba de chi-cuadrado
chi2, p, dof, expected = chi2_contingency(tabla_contingencia_prop)

# Imprimir los resultados
print(f"Valor chi-cuadrado: {chi2}")
print(f"P-valor: {p}")
print(f"Grados de libertad: {dof}")

# Interpretar el p-valor
if p < 0.05:
    print("Existe una correlación estadísticamente significativa entre las dos variables.")
else:
    print("No existe una correlación estadísticamente significativa entre las dos variables.")
```

**Medidas de Asociación**  

**Interpretación:**  
- Como el **p-valor es 0.0 (<0.05)**, podemos **rechazar la hipótesis nula**, lo que indica que **existe una relación estadísticamente significativa** entre `ACRM - NIVEL DE RIESGO` y `NIVEL RIESGO ACT ECONOMICA`.  
- A pesar de esto, esta prueba **no mide la fuerza de la relación**, solo indica que existe.

```python
# Graficar diagramas de caja para identificar valores atípicos
int_columns.plot(kind='box', subplots=True, layout=(5, 5), figsize=(12, 12))
plt.tight_layout()
plt.show()
```

Se puede observar que, de hecho, prácticamente todas las variables numéricas tienen outliers. Por este motivo, es necesario tratar estos valores atípicos. Para este fin, calcularemos los rangos intercuartílicos y reemplazaremos los extremos por estos límites máximos y mínimos.

```python
outliers_int = int_columns.copy()

limits = {}
for col in int_columns:
  q1 = outliers_int[col].quantile(0.25)
  q3 = outliers_int[col].quantile(0.75)
  iqr = q3 - q1
  lower_limit = q1 - 1.5 * iqr
  upper_limit = q3 + 1.5 * iqr
  limits[col] = (lower_limit, upper_limit)

# Replace outliers with limits
for col, (lower_limit, upper_limit) in limits.items():
  outliers_int[col] = outliers_int[col].clip(lower=lower_limit, upper=upper_limit)
```

```python
# Graficar diagramas de caja para identificar valores atípicos
outliers_int.plot(kind='box', subplots=True, layout=(5, 5), figsize=(12, 12))
plt.tight_layout()
plt.show()
```

De esta forma, hemos tratado con los datos atípicos de las variables contínuas del conjunto de datos. Como se puede apreciar en las gráficas, ya no se identifican outliers en ninguna de las columnas.
