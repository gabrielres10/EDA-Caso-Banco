```python
# Descargar los datasets de Drive
#!gdown 1NkzavnWM8_WNg1-0LwduLF9mMZ-0H63s > /dev/null 2>&1
#!gdown 1CbFRVODHfcyJurFry9dQhQdDojEB9yl_ > /dev/null 2>&1

# Cargar dataset
df1 = pd.read_csv('Journal Personas.csv', sep=',', encoding='latin-1', low_memory=False)
df2 = pd.read_excel('Journal transaccional.xlsx', sheet_name='Journal transaccional')


df1_unique = df1.drop_duplicates(subset='ID')
df2.rename(columns={'ID Anónima':'ID'}, inplace=True)

#unir los dos dataset por la columna 'ID'
df = pd.merge(df1_unique, df2, on='ID', how='inner')
df = df[df['ACRM - DEBITO-CREDITO'] != 'R']
df = df[df['ACRM - DEBITO-CREDITO'] != 'E']

#clonar dataframe
data = df.copy()

#borrar columnas con valores vacíos
umbral = 0.8
porcentaje_nulos = data.isnull().mean()
columnas_a_eliminar = porcentaje_nulos[porcentaje_nulos > umbral].index
data = data.drop(columns=columnas_a_eliminar)

# Mostrar resultado
print(f"Tamaño del DataFrame: {data.shape}")
```

```python
risk_form = data['ACRM - NIVEL DE RIESGO'].value_counts(normalize=True).rename('Proportion').reset_index()
risk_form.rename(columns={risk_form.columns[0]: 'Nivel de Riesgo', 'Proportion': 'Porcentaje'}, inplace=True)
risk_form['Porcentaje'] = np.round(risk_form['Porcentaje'] * 100, decimals=2)

if risk_form['Nivel de Riesgo'].dtype in ['int64', 'float64']:
    risk_form['Nivel de Riesgo'] = risk_form['Nivel de Riesgo'].astype(int)

palette = {'Medio': 'orange', 'Alto': 'red', 'Bajo': 'green'}
bar = sns.barplot(data=risk_form, x='Nivel de Riesgo', y='Porcentaje', palette=palette)

for i in range(risk_form.shape[0]):
    bar.text(
        i, risk_form.loc[i, 'Porcentaje'] / 3,
        f"{risk_form.loc[i, 'Porcentaje']}%\n{data['ACRM - NIVEL DE RIESGO'].value_counts()[risk_form.loc[i, 'Nivel de Riesgo']]}",
        fontdict=dict(color='white', fontsize=12, weight='bold'),
        horizontalalignment='center'
    )
```

## Depuración de variables redundantes

Vamos a analizar la correlación que tienen las variables para escoger las mejores dentro del conjunto de datos. Esto nos permitirá evitar repetir información, ya que muchas veces algunas columnas son tan dependientes entre sí que prácticamente proporciona la misma información

```python
# Filtrar solo las columnas numéricas
numeric_data_3 = numeric_data.select_dtypes(include=[np.number])

plt.figure(figsize=(25, 20))
sns.heatmap(numeric_data_3.corr(), annot=False, cmap="coolwarm", linewidths=0.5)
plt.show()
```

Antes del análisis de correlación, se ven varios puntos rojos en el gráfico, que representan variables que proporciona prácticamente la misma información. Vamos a emplear un método para eliminar las variables que tengan más de 85% de correlación entre sí, con el fin de eliminar redundancia.

```python
non_redundant_df = numeric_data.copy()

# Copiamos el DataFrame original
non_redundant_df = numeric_data.copy()

# Definir un umbral de correlación (ejemplo: 0.4)
correlation_threshold = 0.4

# Filtramos solo las columnas numéricas
numeric_cols = non_redundant_df.select_dtypes(include=[np.number])

# Calculamos la matriz de correlación
corr_matrix = numeric_cols.corr().abs()

# Seleccionamos las variables con alta correlación
upper_triangle = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))

# Encontramos las columnas a eliminar (si tienen una correlación mayor al umbral)
to_drop = [column for column in upper_triangle.columns if any(upper_triangle[column] > correlation_threshold)]

# Eliminamos las columnas seleccionadas
non_redundant_df = non_redundant_df.drop(columns=to_drop)

# Añadimos la columna 'NIVEL DE RIESGO' al DataFrame resultante si existe
if 'ACRM - NIVEL DE RIESGO' in numeric_data.columns:
    non_redundant_df['NIVEL DE RIESGO'] = numeric_data['ACRM - NIVEL DE RIESGO']
else:
    raise ValueError("La columna 'ACRM - NIVEL DE RIESGO' no existe en el DataFrame")

# Mostramos las variables eliminadas
print("Variables eliminadas por alta correlación:", to_drop)
```

Parece que se han eliminado múltiples columnas debido a su alta correlación. Esto debería verse reflejado en el mapa de calor. A continuación, veamos el resultado

```python
#asegurarse de que solo hayan variables numericas en non_redundant_df
non_redundant_df = non_redundant_df.select_dtypes(include=[np.number])

plt.figure(figsize=(25, 20))
sns.heatmap(non_redundant_df.corr(), annot=False, cmap="coolwarm", linewidths=0.5)
plt.show()
```

Con este gráfico, podemos observar todas las columnas que quedaron después de eliminar las redundancias. Se puede apreciar que ya no hay una cantidad tan grande de variables dependientes entre sí.

No obstante, ambos gráficos poseen unos espacios vacíos en su interior, lo que sugiere que algunas de las variables tienen un solo valor en todos sus registros.
```python
class DummyTransformer(BaseEstimator, TransformerMixin):
  def __init__(self):
    self.categorical_cols = categorical_cols
    categories = tuple(list(translation_dict[col].values()) for col in self.categorical_cols)
    self.encoder = OneHotEncoder(categories=categories)

  def fit(self, X, y=None):
    self.encoder.fit(X[self.categorical_cols])
    return self

  def transform(self, X, y=None):
    # Perform arbitary transformation
    X_transformed = X.copy()
    encoded_data = encoder.transform(X_transformed[self.categorical_cols]).toarray()
    columns = encoder.get_feature_names_out(self.categorical_cols)
    dummies = pd.DataFrame(encoded_data, columns=columns, index=X_transformed.index).astype(int)
    dummies = dummies.drop(columns=[col+"_sin_informacion" for col in self.categorical_cols])
    X_transformed = pd.concat([X_transformed, dummies], axis = 1)
    X_transformed.drop(columns=self.categorical_cols, inplace=True)
    X_transformed.tiene_propiedades = X_transformed.tiene_propiedades.astype(int)
    return X_transformed
```

Creamos un objeto Transformer que haga este proceso por nosotros.

```python
class PCATransformer(BaseEstimator, TransformerMixin):
  def __init__(self, n_components):
    self.numeric_cols = numeric_cols
    self.n_components = n_components

  def fit(self, X, y=None):
    pca = PCA(self.n_components)
    self.pca = pca.fit(X[numeric_cols])
    return self

  def transform(self, X, y=None):
    # Perform arbitary transformation
    X_transformed = X.copy()
    pcs = self.pca.transform(X_transformed[self.numeric_cols])
    pcs = pd.DataFrame(data = pcs, index=X_transformed.index)
    pcs = pcs.add_prefix('PC')
    X_transformed = pd.concat([X_transformed,pcs],axis=1)
    return X_transformed
```

Se ve que la mejor es con "spherical" con 4 clusters. pero de manera analoga a la cantidad de PCA, la cantidad de clusters será decidido por el optimizador



Creamos un objeto Transformer que haga el clustering por nosotros y que dummifique la variable de clustering, puesto que esta será una simple categoría del cluster al que pertenece.

```python
class ClusteringTransformer(BaseEstimator, TransformerMixin):
  def __init__(self, n_clusters, covariance_type):
    self.n_clusters = n_clusters
    self.covariance_type = covariance_type

  def fit(self, X, y=None):
    self.clustering_columns = [x for x in X.columns if x.startswith('PC')]
    self.clustering_columns.append('LDA')
    self.gmm = GaussianMixture(self.n_clusters, covariance_type=self.covariance_type, random_state=0)
    self.gmm.fit(X[self.clustering_columns])
    return self

  def transform(self, X, y=None):
    # Perform arbitary transformation
    X_transformed = X.copy()
    dummie_clustering = pd.get_dummies(pd.Series(self.gmm.predict(X_transformed[self.clustering_columns]), index=X_transformed.index))
    for i in range(self.n_clusters):
      if i not in dummie_clustering.columns:
        dummie_clustering[str(i)] = 0
    dummie_clustering = dummie_clustering.add_prefix('GMM')
    dummie_clustering = dummie_clustering.drop([f'GMM{self.n_clusters-1}'],axis=1)
    X_transformed = pd.concat([X_transformed,dummie_clustering],axis=1)
    return X_transformed
```
